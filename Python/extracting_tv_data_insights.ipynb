```python
# Load the super_bowls dataset
super_bowls = pd.read_csv("datasets/super_bowls.csv")
super_bowls.head()
```

```python
# Load the tv dataset
tv = pd.read_csv("datasets/tv.csv")
tv.head()
```

```python
# Graph the average_us_viewers versus super_bowl
sns.lineplot(data = tv, x = "super_bowl", y = "avg_us_viewers")
plt.title("Average Number of US Viewers for the Super Bowl")
plt.xlabel("Super Bowl")
plt.ylabel("Average US Viewers")
```
