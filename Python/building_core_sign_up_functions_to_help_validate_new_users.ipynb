```python
# Preloaded data for validating email domain.
top_level_domains = [
    ".org",
    ".net",
    ".edu",
    ".ac",
    ".gov",
    ".com",
    ".io"
]
```

```python
# Implement a function called validate_name()
def validate_name(name: str) -> bool:
    # Check for 'name' data type
    name_dtype = type(name)
    if name_dtype == str:
        # Check for number of characters
        name_len = len(name)
        if name_len <= 2:
            return False
        else:
            return True
    else:
        return False
```

```python
# Implement a function called validate_email()
def validate_email(email: str) -> bool:
    # Split email at '.'
    l = email.split('.')
    # Add the '.' in front of the domain
    domain = '.' + l[1]
    if '@' in email:
        if domain in top_level_domains:
            return True
        else:
            return False
    else:
        return False
```
