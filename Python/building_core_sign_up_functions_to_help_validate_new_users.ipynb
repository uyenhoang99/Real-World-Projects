```python
# Preloaded data for validating email domain.
top_level_domains = [
    ".org",
    ".net",
    ".edu",
    ".ac",
    ".gov",
    ".com",
    ".io"
]
```

```python
# Implement a function called validate_name()
def validate_name(name: str) -> bool:
    # Check for 'name' data type
    name_dtype = type(name)
    if name_dtype == str:
        # Check for number of characters
        name_len = len(name)
        if name_len <= 2:
            return False
        else:
            return True
    else:
        return False
```

```python
# Implement a function called validate_email()
def validate_email(email: str) -> bool:
    # Check for string data type
    email_dtype = type(email)
    if email_dtype == str:
        # Check for '@' symbol
        if '@' in email:
            # Split email at '.'
            email_split = email.split('.')
            # Add the '.' back to the splitted email domain
            email_split = '.' + str(email_split[-1])
            # Check for valid top-level domains
            if email_split in top_level_domains:
                return True
            else:
                return False
        else:
            return False
    else:
        return False
```
